## 2.2 Spatial Regression

When the dependent variable or the errors are spatially dependent, **spatial econometric models** extend the linear regression framework. This section covers: (1) the **classic global models** (OLS, SAR, SEM, SAC, SDM) and **LM-based model selection**; (2) the **position of GWR and MGWR** as local/multiscale tools; and (3) the **calculation methods** for GWR and MGWR.

### 2.2.1 Classic Spatial Regression Models

Notation: $y$ = $n\times 1$ dependent variable, $\mathbf{X}$ = $n\times k$ regressors, $\mathbf{W}$ = $n\times n$ spatial weight matrix (row-standardized), $\varepsilon$ = i.i.d. errors.

| Model | Equation | Interpretation |
|-------|----------|----------------|
| **OLS** | $y = \mathbf{X}\beta + \varepsilon$ | No spatial structure. |
| **SAR / SLM** (Spatial Lag) | $y = \rho \mathbf{W}y + \mathbf{X}\beta + \varepsilon$ | Spillover: $y$ depends on neighbors’ $y$. |
| **SEM** (Spatial Error) | $y = \mathbf{X}\beta + u,\quad u = \lambda \mathbf{W}u + \varepsilon$ | Spatial dependence in errors. |
| **SAC / SARAR** | $y = \rho \mathbf{W}y + \mathbf{X}\beta + u,\quad u = \lambda \mathbf{W}u + \varepsilon$ | Both lag and error dependence. |
| **SDM** (Spatial Durbin) | $y = \rho \mathbf{W}y + \mathbf{X}\beta + \mathbf{W}\mathbf{X}\theta + \varepsilon$ | Lag + spatially lagged regressors. |

- **SAR**: $\rho \neq 0$ → endogenous spillover (neighbors’ outcome affects own outcome).
- **SEM**: $\lambda \neq 0$ → spatial correlation only in errors (e.g. omitted spatially correlated variables).
- **SAC**: allows both $\rho$ and $\lambda$; useful when theory does not rule out either.
- **SDM**: adds $\mathbf{W}\mathbf{X}$ so neighbor characteristics directly affect $y$; often used for robustness or when spillovers in covariates are plausible.

**What each model means in practice**

- **OLS (Ordinary Least Squares)**  
  Standard regression: the outcome at each location $i$ depends only on local covariates $\mathbf{X}_i$ and a random error $\varepsilon_i$. Observations are treated as independent. Use OLS only when there is no evidence of spatial dependence (e.g. LM tests are insignificant). If spatial dependence exists, OLS estimates can be biased or inefficient.

- **SAR / SLM (Spatial Lag Model)**  
  The term $\rho \mathbf{W}y$ means that the *outcome* at location $i$ is influenced by the *outcomes* at neighboring locations. So $y_i$ depends on $y_j$ for neighbors $j$. This is **endogenous spillover** or **neighborhood effects**: e.g. house prices depend on nearby prices; crime in one area depends on crime next door. $\rho > 0$ usually means “similar values cluster in space.” The coefficients $\beta$ give *direct* effects only; total and indirect effects require the spatial multiplier $(\mathbf{I} - \rho \mathbf{W})^{-1}$.

- **SEM (Spatial Error Model)**  
  The *mean* of $y$ is still $\mathbf{X}\beta$ (no spillover in the outcome), but the *errors* $u$ are spatially correlated: $u = \lambda \mathbf{W}u + \varepsilon$. So locations that are close tend to have similar unobserved shocks. This typically reflects **omitted variables that are spatially correlated** (e.g. unmeasured amenities, common shocks) or **measurement error** that is correlated across space. OLS on the same equation is still unbiased for $\beta$, but standard errors are wrong; SEM corrects inference and gives efficient estimates when the error process is spatial.

- **SAC / SARAR (Spatial Autoregressive Combined)**  
  Combines both a **spatial lag** in $y$ and **spatial correlation in the errors** $u$. You allow (i) spillover in the outcome ($\rho \mathbf{W}y$) and (ii) spatial dependence in the residual ($u = \lambda \mathbf{W}u + \varepsilon$). Use SAC when theory or tests suggest both mechanisms might be present, or when you want a flexible specification and let the data choose the relative strength of $\rho$ and $\lambda$.

- **SDM (Spatial Durbin Model)**  
  Extends the lag model by adding **spatially lagged regressors** $\mathbf{W}\mathbf{X}\theta$. So $y$ depends on (i) own and neighbors’ outcomes ($\rho \mathbf{W}y$), (ii) own covariates $\mathbf{X}\beta$, and (iii) *neighbors’ covariates* $\mathbf{W}\mathbf{X}\theta$. That allows “context” or “neighbor characteristics” to affect $y$ directly (e.g. neighbor income). SDM is often used for **robustness** (it nests SAR and can approximate SEM under restrictions) and when you care about distinguishing *direct* vs *indirect* (spatial) effects of covariates.

### 2.2.2 Lagrange Multiplier (LM) Tests for Spatial Dependence

After estimating **OLS** $ \hat{\beta} = (\mathbf{X}'\mathbf{X})^{-1}\mathbf{X}'y $, $ \hat{\varepsilon} = y - \mathbf{X}\hat{\beta} $, we can test for spatial dependence without estimating SAR or SEM.

**Definitions** (with $ \mathbf{W} $ row-standardized, $ \hat{\sigma}^2 = \hat{\varepsilon}'\hat{\varepsilon}/n $, $ \mathbf{M}=\mathbf{I}-\mathbf{X}(\mathbf{X}'\mathbf{X})^{-1}\mathbf{X}' $):

- $ J_\rho = \frac{1}{\hat{\sigma}^2}(\hat{\varepsilon}'\mathbf{W}y)^2 \Big/ \Bigl[ (\mathbf{W}\mathbf{X}\hat{\beta})'\mathbf{M}(\mathbf{W}\mathbf{X}\hat{\beta})/\hat{\sigma}^2 + \mathrm{tr}(\mathbf{W}'\mathbf{W}+\mathbf{W}^2) \Bigr] $  
  → **LM test for lag (LM-$\rho$)**: $ H_0: \rho=0 $. Under $ H_0 $, $ J_\rho \sim \chi^2(1) $.

- $ J_\lambda = \frac{1}{\hat{\sigma}^2}(\hat{\varepsilon}'\mathbf{W}\hat{\varepsilon})^2 \Big/ \mathrm{tr}(\mathbf{W}^2+\mathbf{W}'\mathbf{W}) $  
  → **LM test for error (LM-$\lambda$)**: $ H_0: \lambda=0 $. Under $ H_0 $, $ J_\lambda \sim \chi^2(1) $.

**Robust versions** (so that one test is robust to the other alternative):

- **Robust LM-lag (LM-$\rho^*$)**: tests $ \rho=0 $ when $ \lambda $ may be non-zero.  
  Uses a corrected denominator so that the limit distribution is still $ \chi^2(1) $ under $ H_0: \rho=0 $.

- **Robust LM-error (LM-$\lambda^*$)**: tests $ \lambda=0 $ when $ \rho $ may be non-zero.  
  Similarly corrected so that under $ H_0: \lambda=0 $ the statistic is $ \chi^2(1) $.

(Exact expressions involve the same ingredients $ \hat{\varepsilon} $, $ \mathbf{W}y $, $ \mathbf{W}\hat{\varepsilon} $, $ \mathbf{W}\mathbf{X}\hat{\beta} $, and the trace terms; see Anselin 1988, Anselin et al. 1996.)

### 2.2.3 Model Selection Strategy (LM-based)

A standard **decision rule** (Anselin 1988; Anselin et al. 1996) is:

1. **Estimate OLS**, then compute **LM-$\rho$**, **LM-$\lambda$**, **Robust LM-$\rho^*$**, **Robust LM-$\lambda^*$**.

2. **If both LM-$\rho$ and LM-$\lambda$ are insignificant** (e.g. $p > 0.05$)  
   → Keep **OLS** (no evidence of spatial dependence in lag or error).

3. **If only the lag side is significant**  
   (LM-$\rho$ or Robust LM-$\rho^*$ significant, LM-$\lambda$ and Robust LM-$\lambda^*$ not)  
   → Prefer **SAR (Spatial Lag)**.

4. **If only the error side is significant**  
   (LM-$\lambda$ or Robust LM-$\lambda^*$ significant, LM-$\rho$ and Robust LM-$\rho^*$ not)  
   → Prefer **SEM (Spatial Error)**.

5. **If both sides are significant**  
   (both LM-$\rho$ and LM-$\lambda$ significant)  
   → Use **Robust LM-$\rho^*$** and **Robust LM-$\lambda^*$** to decide:
   - If only Robust LM-$\rho^*$ is significant → **SAR**.
   - If only Robust LM-$\lambda^*$ is significant → **SEM**.
   - If both robust LMs are significant → consider **SAC (SARAR)** or **SDM**, and compare fit (e.g. log-likelihood, AIC/BIC).

6. **SAC vs SDM**: SAC is a flexible baseline when both lag and error dependence are present; SDM is preferred when theory or robustness checks suggest that **neighbor covariates** $\mathbf{W}\mathbf{X}$ should enter the model.

### 2.2.4 Summary

| Step | Action |
|------|--------|
| 1 | Run OLS, then LM-$\rho$, LM-$\lambda$, Robust LM-$\rho^*$, Robust LM-$\lambda^*$. |
| 2 | Both LM insignificant → **OLS**. |
| 3 | Only lag LM(s) significant → **SAR**. |
| 4 | Only error LM(s) significant → **SEM**. |
| 5 | Both significant → use robust LMs; if both robust significant → **SAC** or **SDM**. |

This LM-based approach avoids estimating multiple spatial models before knowing which form of dependence is supported by the data, and is the **classic** way to choose among OLS, SAR, SEM, and SAC in applied spatial econometrics. For **local** coefficient variation (GWR/MGWR), see 2.2.5–2.2.7.

### 2.2.5 GWR and MGWR: where they sit

The models above (OLS, SAR, SEM, SAC, SDM) are **global**: one set of parameters (e.g. one $\beta$) applies to the whole study area. They capture **spatial dependence** (spillovers, correlated errors) but assume the *relationship* between $y$ and $\mathbf{X}$ is the same everywhere.

**GWR (Geographically Weighted Regression)** and **MGWR (Multiscale GWR)** belong to a different family: they allow the **relationship** to change over space (spatial heterogeneity).

| Approach | What it does | When to use |
|----------|----------------|-------------|
| **Global (OLS, SAR, SEM, SAC, SDM)** | One $\beta$ (or one $\rho$, $\lambda$) for all locations; model *dependence* via $\mathbf{W}$. | You care about spillovers, error correlation, or a single “average” effect. |
| **GWR** | $\beta$ varies by location: $y_i = \mathbf{X}_i \beta(u_i,v_i) + \varepsilon_i$. Coefficients are *local* (estimated with distance-based weights). | You suspect the effect of $\mathbf{X}$ on $y$ differs across space (e.g. income effect stronger in center than suburbs). |
| **MGWR** | Like GWR, but **each covariate** can have its own **bandwidth** (spatial scale). So one variable may vary at a very local scale, another at a regional scale. | You want to allow both “global” and “local” drivers in the same model. |

**Position in the workflow**

- **First** use OLS + LM tests to decide whether you need a global spatial model (SAR/SEM/SAC/SDM) at all.
- **Then**, if you are also interested in *varying* effects over space (not just dependence), add **GWR** or **MGWR**:
  - **GWR**: one bandwidth for all coefficients; all effects are local but at the same spatial scale.
  - **MGWR**: different bandwidth per variable; some effects can be almost global (large bandwidth), others very local (small bandwidth).

So: **global models (SAR, SEM, SAC, SDM)** answer “is there spillover or spatial correlation, and in what form?” **GWR/MGWR** answer “does the *effect* of $\mathbf{X}$ on $y$ change from place to place (and at what scale)?” The two can be combined in practice (e.g. fit a global spatial model for dependence, then use GWR/MGWR to explore heterogeneity in effects). Subsections **2.2.6** and **2.2.7** give the **estimation formulas and algorithms** for GWR and MGWR.

### 2.2.6 GWR: calculation method

**Model**  
At each location $i$ with coordinates $(u_i, v_i)$, the model is  
$y_i = \sum_{k=0}^{K} X_{ik}\, \beta_k(u_i, v_i) + \varepsilon_i.$  
So the coefficients $\beta_k$ depend on place; there are $n \times (K+1)$ local parameters (intercept + $K$ regressors).

**Spatial weights**  
For location $i$, observations $j$ are weighted by distance $d_{ij}$ (e.g. Euclidean). Define a **kernel** $k(\cdot)$ and **bandwidth** $b$ (or $b_i$ if adaptive):  
$w_{ij} = k\left( \frac{d_{ij}}{b} \right).$  
(Here $k$ is the kernel function; $K$ above is the number of regressors.) Common choices:  
- **Gaussian**: $k(z) = \exp(-z^2/2)$.  
- **Bisquare**: $k(z) = (1 - z^2)^2$ if $z \leq 1$, else $0$.  
- **Fixed bandwidth** $b$: same $b$ for all $i$.  
- **Adaptive bandwidth**: for each $i$, $b$ is chosen so that the same number of neighbors (e.g. nearest $m$ points) get non-zero weight.

**Local estimate at $i$**  
Stack $y = (y_1,\ldots,y_n)'$, $\mathbf{X}$ = $n \times (K+1)$ (with constant). Let $\mathbf{W}_i = \mathrm{diag}(w_{i1},\ldots,w_{in})$. The **GWR estimate** at $i$ is the weighted least squares solution using $\mathbf{W}_i$:  
$\hat{\beta}(u_i,v_i) = (\mathbf{X}'\mathbf{W}_i\mathbf{X})^{-1} \mathbf{X}'\mathbf{W}_i\, y.$  
So at each $i$ you solve a WLS problem; the fitted value at $i$ is $\hat{y}_i = \mathbf{X}_i' \hat{\beta}(u_i,v_i)$.

**Bandwidth selection**  
- **Cross-validation (CV)**: choose $b$ to minimize $\sum_i (y_i - \hat{y}_{i(-)})^2$, where $\hat{y}_{i(-)}$ is the fitted value at $i$ when $i$ is left out (or down-weighted).
- **AICc**: corrected AIC for small samples; minimize $\mathrm{AICc}(b)$ over $b$.
- **Golden section search** or grid search over $b$ (or over $m$ for adaptive).

**Output**  
For each location $i$ you get $\hat{\beta}_0(u_i,v_i),\ldots,\hat{\beta}_K(u_i,v_i)$. These can be mapped to show how each effect varies over space.

### 2.2.7 MGWR: calculation method

**Idea**  
In GWR, one bandwidth $b$ (or one adaptive $m$) is used for all coefficients, so every $\beta_k$ varies at the **same** spatial scale. In **MGWR**, each coefficient $\beta_k$ has its own bandwidth $b_k$ (or adaptive $m_k$). So one variable can be almost “global” (large $b_k$, slowly varying) and another very “local” (small $b_k$, fast varying).

**Model**  
\[
y_i = \sum_{k=0}^{K} X_{ik}\, \beta_k(u_i, v_i; b_k) + \varepsilon_i,
\]
where $\beta_k(\cdot; b_k)$ is estimated using bandwidth $b_k$ (and hence its own spatial weight matrix $\mathbf{W}^{(k)}$).

**Weights per variable**  
For covariate $k$, at location $i$ define $\mathbf{W}_i^{(k)} = \mathrm{diag}(w_{i1}^{(k)},\ldots,w_{in}^{(k)})$ with $w_{ij}^{(k)} = k(d_{ij}/b_k)$. So the “smoothing” of $\beta_k$ is controlled only by $b_k$.

**Estimation: backfitting**  
MGWR is fitted by **iteratively** updating each coefficient surface given the others:

1. **Initialize**  
   e.g. set all $\hat{\beta}_k$ to the OLS (global) estimate, or to GWR estimates with a common bandwidth.

2. **Update one coefficient**  
   For $k = 0,1,\ldots,K$, hold all $\hat{\beta}_{\ell}$, $\ell \neq k$, fixed and compute the partial residual
   \[
   r_i^{(k)} = y_i - \sum_{\ell \neq k} X_{i\ell}\, \hat{\beta}_\ell(u_i,v_i).
   \]
   Regress $r^{(k)}$ on the $k$-th column of $\mathbf{X}$ using **only** the weights $\mathbf{W}^{(k)}$ (bandwidth $b_k$). That gives an updated $\hat{\beta}_k(u_i,v_i)$ for all $i$.

3. **Cycle**  
   Repeat step 2 for $k=0,1,\ldots,K$ in turn, then repeat the whole cycle until the fitted values (or the coefficient surfaces) change little.

4. **Bandwidth selection per $k$**  
   For each $k$, $b_k$ is chosen by the same criterion as in GWR (e.g. AICc or CV), but **conditional on** the current other $\hat{\beta}_{\ell}$. So in practice, bandwidths and coefficients are updated together (e.g. inner loop: optimize $b_k$ given others; outer loop: backfitting).

**Summary**  
- **GWR**: one $b$ → one weight matrix per location $\mathbf{W}_i$ → $\hat{\beta}(u_i,v_i) = (\mathbf{X}'\mathbf{W}_i\mathbf{X})^{-1}\mathbf{X}'\mathbf{W}_i y$ at each $i$; bandwidth chosen by CV or AICc.  
- **MGWR**: one $b_k$ per covariate $k$ → one $\mathbf{W}^{(k)}$ per $k$ → backfitting over $k$ until convergence; each $b_k$ chosen (e.g. AICc) so that the spatial scale of $\beta_k$ can differ (global vs local).

### 2.2.8 Beyond GWR/MGWR: recent directions

After GWR and MGWR, several lines of work extend local/multiscale modelling:

- **GTWR (Geographically and Temporally Weighted Regression)**  
  Adds **time**: coefficients depend on $(u_i, v_i, t_i)$. Weights use a space–time distance (e.g. combined spatial and temporal bandwidths). Used when relationships vary over both space and time (e.g. panel or repeated cross-sections).

- **Bayesian GWR / GGWR**  
  **Generalized GWR** in a **modularized Bayesian framework** (e.g. Liu & Goudie, *Bayesian Analysis*, 2024) gives proper uncertainty quantification and extends to non-Gaussian outcomes. **Bayesian cluster GWR** combines GWR with mixture models to let effect surfaces cluster into regions (some covariates matter only in certain areas).

- **GNNWR / GTNNWR (Neural-network weighted regression)**  
  **Geographically** (and **Temporally**) **Neural Network Weighted Regression** uses neural networks to learn spatial (and temporal) nonstationarity instead of kernel-weighted least squares. Implementations (e.g. GNNWR Python package) show strong performance when relationships are highly nonlinear or vary in complex ways.

- **Deep learning and vision transformers**  
  **Self-attention** and **vision transformers** have been used to model spatial nonstationarity in gridded or image-like data, often outperforming standard CNNs when spatial structure varies at multiple scales. More suitable when the focus is prediction and the data are high-dimensional (e.g. remote sensing).

In short: **GTWR** adds time; **Bayesian GWR/GGWR** add inference and clustering; **GNNWR/GTNNWR** and **transformers** use neural nets for flexible, data-driven local structure. Choice depends on whether you need interpretable bandwidths (MGWR), uncertainty (Bayesian), time (GTWR), or maximum flexibility (neural approaches).

```python
# Example: OLS + LM tests for spatial dependence (using spreg)
# pip install spreg libpysal
import numpy as np
from libpysal.weights import lat2W
import spreg

# Example data: 5x5 grid
np.random.seed(42)
n = 25
y = np.random.randn(n).cumsum() + 2 * np.arange(n) / n
X = np.column_stack([np.ones(n), np.arange(n) / n])
w = lat2W(5, 5)

# OLS
ols = spreg.OLS(y, X, name_y=["y"], name_x=["const", "x"])
print("OLS coeffs:", ols.betas.T)

# LM tests: lml (lag), lme (error), rlml (robust lag), rlme (robust error)
lms = spreg.LMtests(ols, w)
print("\nLM tests for spatial dependence:")
print("  LM lag (lml):     stat = {:.4f}, p = {:.4f}".format(lms.lml[0], lms.lml[1]))
print("  LM error (lme):   stat = {:.4f}, p = {:.4f}".format(lms.lme[0], lms.lme[1]))
print("  Robust LM lag:   stat = {:.4f}, p = {:.4f}".format(lms.rlml[0], lms.rlml[1]))
print("  Robust LM error: stat = {:.4f}, p = {:.4f}".format(lms.rlme[0], lms.rlme[1]))
```
